```python
import numpy as np
from matplotlib import pyplot as plt
import scipy as sp
```

```python
data = np.loadtxt("upogib.csv", delimiter=",", dtype=None)
m = np.array([data[x][0] for x in range(len(data))])
F=0.00981*m
x1 = [data[x][1] for x in range(len(data))]
x2 = [data[x][2] for x in range(len(data))]
x=np.linspace(0,20,5)
```

```python
k1, n1 = np.polyfit(F, x1, 1)
k2, n2 = np.polyfit(F, x2, 1)
print(k1, n1, k2, n2)
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x, k1*x+n1, color="black")
plt.plot(F, x1, marker="o", linestyle="", markersize=5, label=r"$Kvadratna$")
plt.plot(x, k2*x+n2, color="black")
plt.plot(F, x2, marker="o", linestyle="", markersize=5, label=r"$Okrogla$")
plt.xlabel(r"$F[N]$", fontsize=16)
plt.legend(prop={"size":15})
plt.grid(1)
plt.ylabel(r'$u_0[mm]$', fontsize=16)
plt.savefig("kvadratna.jpg")
plt.show()
```

```python
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x, k2*x+n2, color="black")
plt.plot(F, x2, marker="o", linestyle="", markersize=5)
plt.xlabel("F[N]")
plt.grid(1)
plt.ylabel(r'$u_0[mm]$')
plt.savefig("okrogla.jpg")
plt.show()
```
