```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd
```

```python
d1 = pd.read_fwf('1..txt')
d1 = d1.to_numpy()
nu = d1[:,0]
U = d1[:, 1]

d2 = pd.read_fwf('1 s cunjo.txt')
d2 = d2.to_numpy()
nu1 = d2[:,0]
U1 = d2[:, 1]
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(nu, U, lw=1, label=r"Prazen resonator")
plt.plot(nu1, U1, lw=1, label=r"Dušenje s cunjo")
plt.xlabel(r"$\nu[Hz]$", fontsize=16)
plt.legend(prop={"size":15})
plt.grid(1)
plt.ylabel(r'$U_0[V]$', fontsize=16)
plt.savefig("resonance.jpg")
plt.show()
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.scatter(x1, U1, s=7, label = r"$306Hz, n_x = 1$")
plt.scatter(x2, U2, s=7, label = r"$448Hz, n_x = 0$")
plt.scatter(x3, U3, s=7, label = r"$604Hz, n_x = 2$")
plt.scatter(x4, U4, s=7, label = r"$904Hz, n_x = 1$")

plt.plot(x1, U1, lw=0.7)
plt.plot(x2, U2, lw=0.7)
plt.plot(x3, U3, lw=0.7)
plt.plot(x4, U4, lw=0.7)
plt.xlabel(r"$x[cm]$", fontsize=16)
plt.legend(prop={"size":15}, loc="upper right")
plt.grid(1)
plt.ylabel(r'$U_0[V]$', fontsize=16)
plt.savefig("razbremenitev.jpg")
plt.show()
```
