```python
import numpy as np
from matplotlib import pyplot as plt
import scipy as sp
```

```python
data = np.loadtxt("tuljava.csv", delimiter=",", dtype=None)
h = np.array([data[x][0] for x in range(len(data))])
U = np.array([data[x][1] for x in range(len(data))])
B = 0.0149 * U
dU = [data[x][2] for x in range(len(data))]
dB = (0.16 + dU/U)*B
x=np.linspace(1,23.2,1000)
print(dB)
```

```python
k1, n1 = np.polyfit(h, B, 1)
a1, b1, c1 = np.polyfit(h, B, 2)
a2, b2, c2, d2 = np.polyfit(h, B, 3)
print([k1, n1], [a1, b1, c1], [a2, b2, c2, d2])
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x, k1*x+n1, color="magenta", label=r"$kx + n$")
plt.plot(x, a1*x**2 + b1*x + c1, color="red", label=r"$ax^2 + bx + c$")
plt.plot(x, a2*x**3 + b2*x**2 + c2*x + d2, color="orange", label=r"$ax^3 + bx^2 + cx + d$")
plt.errorbar(h, B, yerr=dB, marker="o", linestyle="", markersize=5)
plt.xlabel(r"$h[cm]$", fontsize=16)
plt.legend(prop={"size":15})
plt.grid(1)
plt.ylabel(r'$B[mT]$', fontsize=16)
plt.savefig("tuljava.jpg")
plt.show()
```
