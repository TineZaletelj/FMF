```python
import numpy as np
from matplotlib import pyplot as plt
import scipy as sp
```

```python
data = np.loadtxt("em.csv", delimiter=",", dtype=None)
I = np.array([data[x][0] for x in range(len(data))])
U = np.array([data[x][1] for x in range(len(data))])
B = 0.149 * U
dU = [data[x][2] for x in range(len(data))]
dB = (0.16 + dU/U)*B
x=np.linspace(0,5,1000)
print(dB)
```

```python
k1, n1 = np.polyfit(I, B, 1)
a1, b1, c1 = np.polyfit(I, B, 2)
a2, b2, c2, d2 = np.polyfit(I, B, 3)
print(k1, n1)
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x, k1*x+n1, color="black")
plt.errorbar(I, B, yerr=dB, marker="o", linestyle="", markersize=5)
plt.xlabel(r"$I[A]$", fontsize=16)
plt.grid(1)
plt.ylabel(r'$B[mT]$', fontsize=16)
plt.savefig("em.jpg")
plt.show()
```
