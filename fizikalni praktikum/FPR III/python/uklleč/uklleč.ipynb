```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd
```

```python
d1 = pd.read_fwf('prečno.txt')
d1 = d1.to_numpy()
U1 = d1[:, 0]
x1 = (np.linspace(0, len(U1), len(U1)) - 23)/5

d2 = pd.read_fwf('prečno_zam.txt')
d2 = d2.to_numpy()
U2 = d2[:, 0]
x2 =( np.linspace(0, len(U2), len(U2)) -13) /5 
```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x1, U1, lw=1, label=r"Poravnan oddajnik")
plt.plot(x2, U2, lw=1, label=r"Zamaknjen oddajnik")
plt.xlabel(r"$x_0[cm]$", fontsize=16)
plt.legend(prop={"size":15})
plt.grid(1)
plt.ylabel(r'$U_0[V]$', fontsize=16)
plt.savefig("precno.jpg")
plt.show()
```

```python
d11 = pd.read_fwf('vzdolž.txt')
d11 = d11.to_numpy()
U11 = d11[:, 0]
x11 = (np.linspace(0, len(U11), len(U11)) - 23)/5 -4

d22 = pd.read_fwf('vzdolž_zam.txt')
d22 = d22.to_numpy()
U22 = 1 - d22[:, 0]
U22 = U22[::-1]
x22 =(np.linspace(0, len(U22), len(U22)) -23) /5 -4

```

```python
plt.rcParams['text.usetex'] = True
plt.figure(figsize=(8,6), dpi=300)
plt.plot(x11, U11, lw=1, label=r"Poravnan oddajnik")
plt.plot(x22, U22, lw=1, label=r"Zamaknjen oddajnik")
plt.xlabel(r"$x_0[cm]$", fontsize=16)
plt.legend(prop={"size":15})
plt.grid(1)
plt.ylabel(r'$U_0[V]$', fontsize=16)
plt.savefig("vzdolz.jpg")
plt.show()
```
