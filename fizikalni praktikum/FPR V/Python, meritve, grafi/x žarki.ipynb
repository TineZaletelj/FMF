```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Nastavitve za LaTeX videz
# -------------------------
plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 12,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.figsize": (5.5, 3.5),   # primerno za \columnwidth
    "lines.linewidth": 1.5,
    "axes.grid": True,
})

# -------------------------
# Podatki
# -------------------------
Uc = np.array([0, 19.8, 41.5, 60.8, 81.8, 102.9, 151.7, 199.4, 249.7, 301.9])
Ue_35 = np.array([0, 0.96, 2.04, 2.8, 3.21, 3.31, 3.4, 3.42, 3.38, 3.38])
Ue_30 = np.array([0, 0.81, 1.58, 1.94, 2.12, 2.14, 2.13, 2.11, 2.1, 2.05])
Ue_25 = np.array([0, 0.6, 1.02, 1.16, 1.21, 1.2, 1.2, 1.07, 1.02, 1.08])

# 5 % napaka
err_35 = 0.05 * Ue_35
err_30 = 0.05 * Ue_30
err_25 = 0.05 * Ue_25

# -------------------------
# Risanje grafa
# -------------------------
plt.figure()

plt.errorbar(Uc, Ue_35, yerr=err_35, fmt='o', capsize=3, markersize=5, label=r'$U_e$ (35 kV)')
plt.errorbar(Uc, Ue_30, yerr=err_30, fmt='o', capsize=3, markersize=5, label=r'$U_e$ (30 kV)')
plt.errorbar(Uc, Ue_25, yerr=err_25, fmt='o', capsize=3, markersize=5, label=r'$U_e$ (25 kV)')

plt.xlabel(r'$U_c$')
plt.ylabel(r'$U_e$')
plt.legend()
plt.tight_layout()

# -------------------------
# Shranjevanje
# -------------------------
plt.savefig("Uc_vs_Ue.pdf")   # priporočeno za LaTeX


plt.show()
```
