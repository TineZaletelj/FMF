```python
import numpy as np
import matplotlib.pyplot as plt

# Nastavitve za LaTeX-style izris
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.size": 12,
    "axes.labelsize": 12,
    "axes.titlesize": 13,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.figsize": (6, 4)
})

# Podatki
t = np.array([0, 6.5, 12, 22, 33, 40, 51, 64, 83, 90, 108, 122])
y = np.array([29.6, 25.7, 23.5, 21.3, 17.8, 16.2, 14.8, 13.5, 11.8, 11.2, 10.5, 10.0])
dy = 0.5  # napaka v y

# Izračun 1/y^2 in napake
inv_y2 = 1 / y**2
d_inv_y2 = 2 * dy / y**3

# Tehtani linearni fit
weights = 1 / d_inv_y2**2
a, b = np.polyfit(t, inv_y2, 1)
fit_line = a * t + b

# Izris grafa
fig, ax = plt.subplots()
ax.errorbar(t, inv_y2, yerr=d_inv_y2, fmt='o', elinewidth=1, capsize=3, label=r'Meritve')
ax.plot(t, fit_line, color='blue', label='Fit')

ax.set_xlabel(r'$t[min]$')
ax.set_ylabel(r'$1/Y_{max}^2[cm^{-2}]$')
ax.set_title(r'Graf $1/Y_{max}^2$ v odvisnosti od časa')
ax.legend()
ax.grid(True, which='both', linestyle=':', linewidth=0.6)

plt.tight_layout()
plt.savefig("inv_y2_vs_t.pdf", bbox_inches='tight')  # za LaTeX vključitev
plt.show()

print(f"Fit koeficienti:\na = {a:.6e}\nb = {b:.6e}")
```
