```python
import numpy as np
from scipy.optimize import curve_fit
from matplotlib import pyplot as plt

phi=[69.8, 70.1, 71.4, 69.4, 70.8, 71.5]
lam=[577, 546, 436, 656, 486, 434]

hgx=[69.8, 70.1, 71.4]
hgy=[577, 546, 436]

h2x=[69.4, 70.8, 71.5]
h2y=[656, 486, 434]

def test(x, a,b,c):
    return a + b*x + c*np.sqrt(x)

d=np.linspace(430,700,10000)

p,n=curve_fit(test, lam, phi)

a=p[0]
b=p[1]
c=p[2]
print(a,b,c,n)

plt.rc("text", usetex=True)
plt.errorbar(hgy, hgx, yerr=0.1, fmt="o", c="r", label=r"$Hg$")
plt.errorbar(h2y, h2x, yerr=0.1, fmt="o", c="b", label=r"$H_2$")
plt.plot(d,test(d,a,b,c), label="fit")
plt.legend()
plt.xlabel(r"$\lambda[nm]$")
plt.ylabel(r"$\varphi[^\circ]$")
plt.show()


f = lambda x: (c + np.sqrt(c**2-4*b*(a-x)))/(2*b)

print(f(69.4)**2)
```
